```python
cutflow_array = []
for pdi in lists.keys():
    cutflow=load('../hists/darkhiggs2018/cutflow--'+pdi+'.reduced')
    sumw=load('../hists/darkhiggs2018/sumw--'+pdi+'.reduced')
    
    scale=sumw['sumw'].integrate('dataset', pdi).values(overflow='all')[()][1]
    num=cutflow['cutflow'].integrate('region','sr').integrate('dataset',pdi).values()[()][12,1]/scale*xsec[pdi]
    den=cutflow['cutflow'].integrate('region','sr').integrate('dataset',pdi).integrate('ZHbbvsQCD').values()[()][0]/scale*xsec[pdi]
    print(pdi,num/den)
    if 'Mz1000_mhs130_Mdm150' in pdi:
        cutflow_array=(cutflow['cutflow'].integrate('region','sr').integrate('dataset',pdi).integrate('ZHbbvsQCD').values()[()][:13]/scale*xsec[pdi]).tolist()
        cutflow_array.append(num)
```

```python
cuts = ['original',
        'met_filters',
        'noHEMj',
        'noHEMmet',
        'met_triggers',
        'recoil_sr',
        'calo_sr',
        'fatjet',
        'msd',
        'mindphi_sr',
        'minDphi_sr',
        'iszeroL',
        'noextrab',
        'deepAK15']
for cut in cuts:
    print(cut,cutflow_array[cuts.index(cut)])
```
